```python
nums = input("Enter the numbers separated by spaces").split(" ")
nums = [int(i) for i in nums]
```

```python
import numpy as np
```

```python
leng = len(nums)
s = sum(nums)
avg = round(s/leng, 3)
avg_nums = [abs(p) for p in nums]
x = np.abs(np.array(nums) - np.array([avg for i in nums]))
print("Average = ", avg)
print("Absolute Average = ", round(sum(list(map(lambda x: abs(x), nums)))/leng, 3))
print("List = ", nums)
print("Absolute Errors = ", x)
print("Mean absolute Error = ", x.mean())
```

```python
salaries = np.array([25, 30, 28, 27, 29, 31, 100, 5])
mean_salary = np.mean(salaries)
```

```python
deviation = np.abs(salaries - mean_salary)
thresholds1 = 1.5*np.mean(deviation)
thresholds2 = 3*np.mean(deviation)
```

```python
outliers1 = salaries[deviation > thresholds1]
outliers2 = salaries[deviation > thresholds2]
```

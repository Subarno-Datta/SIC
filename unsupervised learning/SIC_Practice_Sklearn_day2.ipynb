# **K FOLD CROSS VALIDATION**


## The dataset is broken into let's say, 3 parts..

#### First iteration, the first part is used for testing, whereas the other 2 parts are used for training..

#### Second iteration, second part is used for testing, whereas the other 2 parts are used for training..

#### Third iteration, third part is used for testing, whereas the other 2 parts are used for testing..

#### Thus, number of folds = number of partitions

```python
cancer_df = pd.concat([feature_df, target_df], axis=1)
```

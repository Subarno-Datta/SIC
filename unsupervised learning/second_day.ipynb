```python
from sklearn.datasets import load_iris
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
%matplotlib inline
```

```python
iris = load_iris()
```

```python
iris
```

```python
iris_df = pd.DataFrame(data=iris['data'])
```

```python
iris_df.columns = iris['feature_names']
```

```python
iris_df.iloc[:,2:4]
```

```python
iris_df.iloc[0]
```

```python
iris_df
```

```python
from sklearn.cluster import AgglomerativeClustering
likage = ['complete', 'average', 'ward']
for idx, i in enumerate(likage):
    plt.figure(idx)
    hier = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage=i)
    hier.fit(iris_df.iloc[:, 2:4])
    plt.scatter(iris_df.iloc[:,2], iris_df.iloc[:,3], c=hier.labels_)
    plt.title('Clustering : ' + i)
    plt.xlabel('petal length')
    plt.ylabel('petal width')
    plt.show()
```

```python
from scipy.cluster import hierarchy
```

```python
heirar = hierarchy.linkage(iris_df.iloc[:,2:4], 'complete')
plt.figure(figsize=(20,5))
dn = hierarchy.dendrogram(heirar)
plt.show()
```

```python
hierar = hierarchy.linkage(iris_df.iloc[:,2:4], 'single')
plt.figure(figsize=(20,10))
dn2 = hierarchy.dendrogram(hierar)
plt.show()
```

```python
import io, urllib.request
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import keras

rng = np.random.default_rng(42)
np.set_printoptions(precision=2, suppress=True, linewidth=120)

mnist = np.load(io.BytesIO(urllib.request.urlopen("https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz").read()))
x_train, y_train = mnist["x_train"] / 255.0, mnist["y_train"]
x_test, y_test = mnist["x_test"] / 255.0, mnist["y_test"]
digit = x_test[0]
print("MNIST:", x_train.shape, x_test.shape)
```

---
# Part 1: the problem — dense layers and images
A dense layer flattens the image into 784 numbers: it has **no idea** which pixels are neighbours. Train a small MLP, then test it on the **same** test digits shifted a few pixels to the right.

```python
def train_mlp(X, y, epochs=3, lr=0.001):
    # 784 -> 128 -> 10, ReLU, Adam, mini-batches of 64 (from the deep-networks folder)
    W = [rng.normal(0, np.sqrt(2 / 784), (784, 128)), rng.normal(0, np.sqrt(2 / 128), (128, 10))]
    B = [np.zeros(128), np.zeros(10)]
    params = W + B
    m, s, t = [np.zeros_like(p) for p in params], [np.zeros_like(p) for p in params], 0
    for _ in range(epochs):
        idx = rng.permutation(len(X))
        for start in range(0, len(X), 64):
            xb, yb = X[idx[start:start + 64]], y[idx[start:start + 64]]
            h = np.maximum(0, xb @ W[0] + B[0])
            S = h @ W[1] + B[1]
            P = np.exp(S - S.max(1, keepdims=True))
            P /= P.sum(1, keepdims=True)
            d2 = (P - np.eye(10)[yb]) / len(yb)
            d1 = (d2 @ W[1].T) * (h > 0)
            grads = [xb.T @ d1, h.T @ d2, d1.sum(0), d2.sum(0)]
            t += 1
            for k, (p, g) in enumerate(zip(params, grads)):
                m[k] = 0.9 * m[k] + 0.1 * g
                s[k] = 0.999 * s[k] + 0.001 * g ** 2
                p -= lr * (m[k] / (1 - 0.9 ** t)) / (np.sqrt(s[k] / (1 - 0.999 ** t)) + 1e-8)
    return lambda X_: (np.maximum(0, X_ @ W[0] + B[0]) @ W[1] + B[1]).argmax(1)

mlp_predict = train_mlp(x_train.reshape(-1, 784), y_train)
print(f"{'shift (pixels right)':>20} | {'MLP test accuracy':>17}")
print("-" * 42)
for shift in [0, 1, 2, 3, 4]:
    shifted = np.roll(x_test, shift, axis=2)
    print(f"{shift:>20} | {100 * np.mean(mlp_predict(shifted.reshape(-1, 784)) == y_test):16.1f}%")

fig, axes = plt.subplots(1, 3, figsize=(7, 2.5))
for ax, shift in zip(axes, [0, 2, 4]):
    ax.imshow(np.roll(x_test[0], shift, axis=1), cmap="gray_r")
    ax.set(title=f"shift {shift}", xticks=[], yticks=[])
plt.show()
```

The same digit moved 4 pixels is still obviously a 7 to us, but the MLP's accuracy collapses: each of its weights is tied to one fixed pixel position.
Second problem: size. A 224×224 colour photo has 150,528 numbers; one dense layer of 1,000 neurons would need **150 million weights**.

# Part 2: why the 2D layout matters — flattening throws it away
An image is a **grid**: every pixel has neighbours left, right, above and below. A dense layer first flattens the grid into one long row. Label each pixel of a 5×5 image with its position in that row:

```python
grid = np.arange(25).reshape(5, 5)
print("5×5 image, each pixel labelled with its position in the flat vector:\n", grid)
r, c = 2, 2
print(f"\npixel (row {r}, col {c}) is position {grid[r, c]}. Its 4 neighbours in the image:")
for name, (dr, dc) in [("left", (0, -1)), ("right", (0, 1)), ("above", (-1, 0)), ("below", (1, 0))]:
    print(f"   {name:>5}: position {grid[r + dr, c + dc]:>2}  ({abs(grid[r + dr, c + dc] - grid[r, c])} apart in the flat vector)")
print(f"\nand positions 4 and 5 sit next to each other in the flat vector, but in the image one is at the right edge of row 0, the other at the left edge of row 1")
```

After flattening, "above" and "below" are 5 positions away, and some pixels that are next to each other in the row are far apart in the image. The dense layer is **never told** which inputs were neighbours.

**Test:** shuffle the 784 pixel positions of every MNIST digit with **one fixed** shuffle. To us the digits become noise. Does the MLP care?

```python
order = rng.permutation(784)                                       # the same shuffle for every image
scrambled_train, scrambled_test = x_train.reshape(-1, 784)[:, order], x_test.reshape(-1, 784)[:, order]
mlp_scrambled = train_mlp(scrambled_train, y_train)

print(f"MLP on normal digits   : {100 * np.mean(mlp_predict(x_test.reshape(-1, 784)) == y_test):.1f}% test accuracy")
print(f"MLP on scrambled digits: {100 * np.mean(mlp_scrambled(scrambled_test) == y_test):.1f}% test accuracy")

fig, axes = plt.subplots(1, 4, figsize=(9, 2.5))
for k in range(2):
    axes[2 * k].imshow(x_test[k], cmap="gray_r")
    axes[2 * k + 1].imshow(scrambled_test[k].reshape(28, 28), cmap="gray_r")
    axes[2 * k].set(title=f"digit {y_test[k]}", xticks=[], yticks=[])
    axes[2 * k + 1].set(title="same, scrambled", xticks=[], yticks=[])
plt.show()
```

**The same accuracy.** The MLP never used the 2D layout at all — it treats the image as a bag of 784 unrelated numbers. That is why it breaks when a digit moves: pixel 300 is just "input 300" to it.

A convolution does the opposite: it looks at a small **3×3 patch of neighbours** and uses the **same** small filter everywhere. So it keeps the 2D layout, needs very few weights, and a pattern is found wherever it is.

# Part 3: forward pass by hand — a 3×3 kernel sliding over a 5×5 image
Put the 3×3 **kernel** (filter) on the top-left 3×3 window, multiply the 9 pairs, add them up — that is one output number. Slide one step, repeat. A 5×5 image has 3 × 3 = 9 window positions.

```python
image = np.array([[1, 2, 0, 1, 3],
                  [0, 1, 3, 2, 1],
                  [2, 0, 1, 0, 2],
                  [1, 3, 2, 1, 0],
                  [0, 1, 0, 2, 1]], dtype=float)
kernel = np.array([[1, 0, -1],
                   [1, 0, -1],
                   [1, 0, -1]], dtype=float)                     # vertical-edge detector: left minus right

def num(v):
    # print -1 as (-1) so the sums are easy to read
    return f"({v:g})" if v < 0 else f"{v:g}"

window = image[0:3, 0:3]
print("top-left window:\n", window, "\nkernel:\n", kernel)
print("\noutput[0,0] = " + " + ".join(f"{num(x)}·{num(w)}" for x, w in zip(window.ravel(), kernel.ravel())) + f" = {np.sum(window * kernel):g}")

out = np.zeros((3, 3))
for r in range(3):
    for c in range(3):
        out[r, c] = np.sum(image[r:r + 3, c:c + 3] * kernel)      # window at (r, c) × kernel, added up
print("\nall 9 window positions (3×3 output):\n", out)

bias = 1.0
z = out + bias
print(f"\n+ bias {bias:g}:\n", z, "\nafter ReLU, max(0, z):\n", np.maximum(0, z))
```

The same thing, fast: build every window at once (`sliding_window_view`), then one multiply-and-sum. We use this `conv2d` for the rest of the notebook.

```python
def conv2d(image, kernel, padding=0, stride=1):
    # image (H, W), kernel (k, k): slide, multiply, add up
    if padding:
        image = np.pad(image, padding)
    k = kernel.shape[0]
    windows = np.lib.stride_tricks.sliding_window_view(image, (k, k))[::stride, ::stride]
    return np.einsum("ijkl,kl->ij", windows, kernel)

print("fast version equal to the loop?", np.allclose(conv2d(image, kernel), out))
```

**Stride** = how far the window jumps. **Padding** = a frame of zeros around the image, so the border pixels get used too and the size can stay the same. Same 5×5 image and kernel, four settings, each checked with the output-size formula $\frac{n + 2p - k}{s} + 1$:

```python
print("image with padding 1 (7×7):\n", np.pad(image, 1), "\n")
for padding, stride in [(0, 1), (0, 2), (1, 1), (1, 2)]:
    result = conv2d(image, kernel, padding=padding, stride=stride)
    size = (5 + 2 * padding - 3) // stride + 1
    corners = [(r, c) for r in range(0, 5 + 2 * padding - 2, stride) for c in range(0, 5 + 2 * padding - 2, stride)]
    print(f"padding {padding}, stride {stride}: (5 + 2·{padding} − 3)/{stride} + 1 = {size} ➜ output {result.shape}")
    print(f"   window top-left corners: {corners}")
    print(result, "\n")

print("the formula on bigger images:")
for n, k, p, s in [(28, 3, 1, 1), (28, 3, 1, 2), (32, 5, 0, 1), (224, 7, 3, 2)]:
    measured = conv2d(rng.random((n, n)), np.ones((k, k)), padding=p, stride=s).shape[0]
    print(f"   n {n:>3}, k {k}, p {p}, s {s}: ({n} + {2 * p} − {k})/{s} + 1 = {(n + 2 * p - k) // s + 1:>3} | measured {measured}")
```

Same numbers from Keras: a `Conv2D` layer with **our** kernel as its weights (`padding="same"` = padding 1 for a 3×3 kernel).

```python
for padding, stride in [(0, 1), (0, 2), (1, 1), (1, 2)]:
    layer = keras.layers.Conv2D(1, 3, strides=stride, padding="same" if padding else "valid")
    layer.build((1, 5, 5, 1))
    layer.set_weights([kernel[:, :, None, None], np.array([bias])])      # weights (3, 3, in 1, out 1), bias (1,)
    keras_out = layer(image[None, :, :, None]).numpy()[0, :, :, 0]
    ours = conv2d(image, kernel, padding=padding, stride=stride) + bias
    print(f"padding {padding}, stride {stride}: Keras output {keras_out.shape} equal to ours? {np.allclose(keras_out, ours)}")
```

# Part 4: image processing filters — what 3×3 kernels do to real photos
Long before deep learning, image editors (Photoshop's blur and sharpen, phone cameras, medical imaging) used exactly this operation with **hand-designed** kernels. Same `conv2d`, three real photos. Every filter below is shown with **its numbers** next to what it does.

```python
PHOTOS = {
    "dog":      "https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg",
    "cat":      "https://storage.googleapis.com/download.tensorflow.org/example_images/320px-Felis_catus-cat_on_snow.jpg",
    "portrait": "https://storage.googleapis.com/download.tensorflow.org/example_images/grace_hopper.jpg",
}
colour, gray = {}, {}
for name, url in PHOTOS.items():
    path = keras.utils.get_file(url.split("/")[-1], url)
    colour[name] = np.array(keras.utils.load_img(path), dtype=float) / 255.0       # (H, W, 3)
    gray[name] = colour[name] @ np.array([0.299, 0.587, 0.114])                    # colour ➜ brightness
    print(f"{name:>8}: {colour[name].shape} ➜ grayscale {gray[name].shape}")

def gaussian(size, sigma):
    x = np.arange(size) - size // 2
    g = np.exp(-x ** 2 / (2 * sigma ** 2))
    return np.outer(g, g) / np.outer(g, g).sum()

KERNELS = {
    "identity":         np.array([[0, 0, 0], [0, 1, 0], [0, 0, 0]], dtype=float),
    "box blur":         np.ones((3, 3)) / 9,
    "Gaussian blur":    np.outer([1, 2, 1], [1, 2, 1]) / 16,
    "sharpen":          np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]], dtype=float),
    "Sobel x":          np.array([[1, 0, -1], [2, 0, -2], [1, 0, -1]], dtype=float),     # vertical edges
    "Sobel y":          np.array([[1, 2, 1], [0, 0, 0], [-1, -2, -1]], dtype=float),     # horizontal edges
    "Laplacian":        np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]], dtype=float),       # edges in every direction
    "emboss":           np.array([[-2, -1, 0], [-1, 1, 1], [0, 1, 2]], dtype=float),
}
EDGE_FILTERS = {"Sobel x", "Sobel y", "Laplacian"}

def show_kernel(ax, k, title):
    m = np.abs(k).max()
    ax.imshow(k, cmap="RdBu_r", vmin=-m, vmax=m)
    for (i, j), v in np.ndenumerate(k):
        ax.text(j, i, f"{v:.2g}", ha="center", va="center", fontsize=10, color="white" if abs(v) > 0.6 * m else "black")
    ax.set(title=title, xticks=[], yticks=[])

def show_result(ax, result, edges, title):
    if edges:
        ax.imshow(np.abs(result), cmap="gray_r")                   # strength of the edge, dark = strong
    else:
        ax.imshow(np.clip(result, 0, 1), cmap="gray")
    ax.set(title=title, xticks=[], yticks=[])

head = gray["dog"][30:230, 20:220]                                # zoom on the head: blur and sharpen are easier to see
fig, axes = plt.subplots(3, len(KERNELS), figsize=(22, 9.5), gridspec_kw={"height_ratios": [1, 2.2, 2.6]})
for k, (name, kern) in enumerate(KERNELS.items()):
    show_kernel(axes[0, k], kern, name)
    show_result(axes[1, k], conv2d(gray["dog"], kern, padding=1), name in EDGE_FILTERS, "")
    show_result(axes[2, k], conv2d(head, kern, padding=1), name in EDGE_FILTERS, "zoom: head")
fig.suptitle("Eight 3×3 kernels on the same photo (top: the 9 numbers, red = positive, blue = negative; bottom: zoomed in)")
plt.tight_layout()
plt.show()
```

How to read a kernel:
- **all positive, adding up to 1** (box, Gaussian) ➜ an **average** of the neighbourhood ➜ **blur**
- **positive on one side, negative on the other, adding up to 0** (Sobel) ➜ "left minus right" ➜ big only where brightness **changes** ➜ **edges** in one direction (flat areas give 0)
- **centre against all neighbours, adding up to 0** (Laplacian) ➜ edges in **every** direction
- **sharpen** adds up to 1: it keeps the photo **and** adds the edges on top. In fact sharpen = identity − Laplacian:

```python
print("sharpen == identity − Laplacian ?", np.allclose(KERNELS["sharpen"], KERNELS["identity"] - KERNELS["Laplacian"]))
for name, kern in KERNELS.items():
    print(f"   {name:>13}: the 9 numbers add up to {kern.sum():g}")
```

## Bigger kernels: more blur
A 3×3 box only averages a tiny neighbourhood. A 9×9 kernel (81 numbers) averages a much bigger one. The Gaussian gives the centre more weight than the edges of the window, so it blurs more smoothly than the box.

```python
blurs = {"original": None, "box 3×3": np.ones((3, 3)) / 9, "box 9×9": np.ones((9, 9)) / 81, "Gaussian 9×9 (σ = 2)": gaussian(9, 2)}
fig, axes = plt.subplots(1, 4, figsize=(16, 3.6))
for ax, (name, kern) in zip(axes, blurs.items()):
    result = gray["cat"] if kern is None else conv2d(gray["cat"], kern, padding=kern.shape[0] // 2)
    show_result(ax, result, False, f"{name}  {result.shape[0]}×{result.shape[1]}")
plt.tight_layout()
plt.show()
```

## Edge strength: two Sobel filters combined
Sobel x finds vertical edges, Sobel y horizontal ones. At each pixel, combine them like the two sides of a right triangle: $\text{strength} = \sqrt{G_x^2 + G_y^2}$ — all edges, in any direction.

```python
gx = conv2d(gray["portrait"], KERNELS["Sobel x"], padding=1)
gy = conv2d(gray["portrait"], KERNELS["Sobel y"], padding=1)
strength = np.sqrt(gx ** 2 + gy ** 2)

fig, axes = plt.subplots(1, 4, figsize=(16, 4.6))
axes[0].imshow(gray["portrait"], cmap="gray")
axes[0].set(title="portrait", xticks=[], yticks=[])
show_result(axes[1], gx, True, "Sobel x: vertical edges")
show_result(axes[2], gy, True, "Sobel y: horizontal edges")
show_result(axes[3], strength, True, "√(Gx² + Gy²): all edges")
plt.tight_layout()
plt.show()
```

## A classic edge detector: blur ➜ edge strength ➜ threshold
Edge filters also react to tiny texture (grass, fur, noise). The classic recipe (the idea behind the Canny detector, 1986): **blur first** to remove the tiny texture, then compute the edge strength, then keep only the strongest 10% as lines. Three versions: no blur, a small blur, a strong blur.

```python
def edge_drawing(img, blur, keep=0.10):
    if blur:                                                       # blur = (kernel size, sigma)
        img = conv2d(img, gaussian(*blur), padding=blur[0] // 2)
    s = np.sqrt(conv2d(img, KERNELS["Sobel x"], padding=1) ** 2 + conv2d(img, KERNELS["Sobel y"], padding=1) ** 2)
    return s > np.quantile(s, 1 - keep)                            # True = one of the strongest 10% of edges

fig, axes = plt.subplots(1, 4, figsize=(20, 4.4))
axes[0].imshow(colour["dog"])
axes[0].set(title="photo", xticks=[], yticks=[])
for ax, blur, title in [(axes[1], None, "no blur"), (axes[2], (5, 1.5), "small blur: Gaussian 5×5, σ 1.5"), (axes[3], (9, 3), "strong blur: Gaussian 9×9, σ 3")]:
    ax.imshow(edge_drawing(gray["dog"], blur), cmap="gray_r")
    ax.set(title=f"{title} ➜ strongest 10%", xticks=[], yticks=[])
plt.tight_layout()
plt.show()
```

- **No blur:** the dog's outline is there, but the grass texture produces thousands of tiny "edges" that fill the bottom half.
- **Small blur:** the outline gets thicker and better connected, but much of the grass is still there.
- **Strong blur:** the grass texture is averaged away and what is left is almost only the **outline** — a line drawing of the dog. The price: thin details (fur, the inside of the ear) disappear too. Choosing the blur is a trade-off.
- The **frame** around the image is not an edge of the dog: **zero padding** puts black pixels next to the bright border, and the edge filter sees that jump (padding, Part 3).

Choosing these steps — which blur, which edge filter, which threshold — was the hand-work of computer vision for decades. A CNN learns its own steps.

## Try your own kernel
Change the 9 numbers and run the cell again. Ideas: make the centre bigger; put all the negatives on top (what direction of edge do you get?); make every number 1/9 but the centre 2; copy Sobel and turn it 45°.

```python
MY_KERNEL = np.array([[ -1, -1,  -1],
                      [-1,  8, -1],
                      [ -1, -1,  -1]], dtype=float)                # ← change these 9 numbers
MY_PHOTO = "cat"                                                   # "dog", "cat" or "portrait"

result = conv2d(gray[MY_PHOTO], MY_KERNEL, padding=1)
fig, axes = plt.subplots(1, 3, figsize=(14, 4), gridspec_kw={"width_ratios": [1, 2.5, 2.5]})
show_kernel(axes[0], MY_KERNEL, f"your kernel (sum {MY_KERNEL.sum():g})")
axes[1].imshow(gray[MY_PHOTO], cmap="gray")
axes[1].set(title="original", xticks=[], yticks=[])
show_result(axes[2], result, abs(MY_KERNEL.sum()) < 1e-9, "your filter (sum 0: shown as edge strength)")
plt.tight_layout()
plt.show()
print(f"output range {result.min():.2f} to {result.max():.2f}")
```

Or with sliders: type numbers into the 3×3 grid, pick a photo, press **apply** (needs Colab or Jupyter — `ipywidgets`).

```python
import ipywidgets as widgets
from IPython.display import display

cells_in = [widgets.FloatText(value=float(v), layout=widgets.Layout(width="70px")) for v in KERNELS["Sobel x"].ravel()]
photo_choice = widgets.Dropdown(options=list(PHOTOS), value="dog", description="photo")
preset = widgets.Dropdown(options=list(KERNELS), value="Sobel x", description="start from")
apply_button = widgets.Button(description="apply", button_style="primary")
output = widgets.Output()

def load_preset(change):
    for box, v in zip(cells_in, KERNELS[change["new"]].ravel()):
        box.value = float(v)

def apply_kernel(_):
    k = np.array([box.value for box in cells_in]).reshape(3, 3)
    with output:
        output.clear_output(wait=True)
        fig, axes = plt.subplots(1, 2, figsize=(11, 4), gridspec_kw={"width_ratios": [1, 2.5]})
        show_kernel(axes[0], k, f"kernel (sum {k.sum():g})")
        show_result(axes[1], conv2d(gray[photo_choice.value], k, padding=1), abs(k.sum()) < 1e-9, photo_choice.value)
        plt.show()

preset.observe(load_preset, names="value")
apply_button.on_click(apply_kernel)
display(widgets.VBox([widgets.HBox([preset, photo_choice]), widgets.GridBox(cells_in, layout=widgets.Layout(grid_template_columns="repeat(3, 80px)")),
                      apply_button, output]))
```

## The bridge to CNNs: 8 filters = one `Conv2D` layer
Put 8 of these kernels into **one** Keras `Conv2D` layer with 8 filters. One call ➜ 8 feature maps, exactly our 8 filters.

```python
names = list(KERNELS)
bank = np.stack([KERNELS[n] for n in names], axis=-1)[:, :, None, :]       # (3, 3, 1 input channel, 8 filters)
conv_layer8 = keras.layers.Conv2D(8, 3, padding="same", use_bias=False)
h, w = gray["dog"].shape
conv_layer8.build((None, h, w, 1))
conv_layer8.set_weights([bank])
feature_maps = conv_layer8(gray["dog"][None, :, :, None]).numpy()[0]       # (H, W, 8)
print(f"input {gray['dog'].shape} ➜ one Conv2D layer, {conv_layer8.count_params()} weights (8 filters × 9) ➜ output {feature_maps.shape}")
print("each feature map == our conv2d with that kernel?",
      all(np.allclose(feature_maps[:, :, k], conv2d(gray["dog"], KERNELS[n], padding=1), atol=1e-4) for k, n in enumerate(names)))

fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, (k, n) in zip(axes.ravel(), enumerate(names)):
    show_result(ax, feature_maps[:, :, k], n in EDGE_FILTERS, f"feature map {k}: {n}")
fig.suptitle("One Conv2D layer with 8 filters ➜ 8 feature maps")
plt.tight_layout()
plt.show()
```

**This is what the first layer of a CNN is**: a bank of small filters, each producing one feature map. The only difference: **we** chose these 72 numbers; a CNN **learns** them from labelled examples. Can a network really find a filter like Sobel by itself? The next part shows the backward pass that makes it possible — and then does exactly that.

---
# Part 5: backward pass by hand, and weight sharing
## First, weight sharing: a convolution is a dense layer with tied weights
The 3×3 output of Part 3 is 9 numbers, each a weighted sum of input pixels — so we can write the convolution as a **9 × 25 dense weight matrix**. Row = one output, column = one of the 25 pixels.

```python
M = np.zeros((9, 25))
for r in range(3):
    for c in range(3):
        placed = np.zeros((5, 5))
        placed[r:r + 3, c:c + 3] = kernel                    # the kernel placed over window (r, c)
        M[3 * r + c] = placed.ravel()
print("the convolution as a 9 × 25 weight matrix:\n", M.astype(int))
print("\nM @ (flat image) == convolution?", np.allclose(M @ image.ravel(), conv2d(image, kernel).ravel()))
print(f"\na dense layer 25 ➜ 9 learns {M.size} separate weights")
print(f"the convolution: {np.sum(M != 0)} non-zero entries in the matrix, but they are copies of only {kernel.size} numbers")
```

Two ideas are visible in one matrix:
- **local connections** — most entries are 0: each output sees only its own 3×3 window
- **weight sharing** — every row uses the **same 9 numbers**, just shifted

So in backprop each kernel weight is used at **all 9 positions**, and its gradient is the **sum of 9 terms**, one per position.

## The backward pass
Forward: $\text{out} = X \circledast W + b$ (3×3), and a simple loss $L = \tfrac12 \sum (\text{out} - \text{target})^2$, so $\delta = \frac{\partial L}{\partial\, \text{out}} = \text{out} - \text{target}$ (also 3×3).

| gradient | formula | in words |
|---|---|---|
| $\frac{\partial L}{\partial W_{ij}}$ | $\sum_{r,c} \delta_{rc}\, X_{r+i,\,c+j}$ | weight $W_{ij}$ touched pixel $X_{r+i,c+j}$ at every position: add them all |
| $\frac{\partial L}{\partial b}$ | $\sum_{r,c} \delta_{rc}$ | the bias is shared by all 9 outputs |
| $\frac{\partial L}{\partial X_{ab}}$ | sum of $\delta_{rc}\, W_{a-r,\,b-c}$ over every window that covered $X_{ab}$ | needed when the layer below also learns |

```python
X, W, b = image, kernel.copy(), 0.0
target = np.array([[0, 2, 0],
                   [1, 0, -1],
                   [0, -2, 0]], dtype=float)                    # the output we want (made up)

out = conv2d(X, W) + b
delta = out - target
print("output:\n", out, "\ntarget:\n", target, "\ndelta = output − target:\n", delta)
print(f"loss = ½ Σ delta² = {0.5 * np.sum(delta ** 2):g}")
```

```python
dW = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        dW[i, j] = sum(delta[r, c] * X[r + i, c + j] for r in range(3) for c in range(3))

print("dL/dW[0,0]: W[0,0] was used at all 9 window positions ➜ 9 terms (delta · the pixel it touched):")
print("   = " + " + ".join(f"{num(delta[r, c])}·{num(X[r, c])}" for r in range(3) for c in range(3)) + f" = {dW[0, 0]:g}")
print("\ndL/dW for all 9 weights:\n", dW)
print("shortcut — the same as convolving the input X with delta:", np.allclose(dW, conv2d(X, delta)))

db = delta.sum()
print(f"\ndL/db = sum of the 9 deltas = {db:g}")
```

```python
dX = np.zeros((5, 5))
used = np.zeros((5, 5), dtype=int)
for r in range(3):
    for c in range(3):
        dX[r:r + 3, c:c + 3] += delta[r, c] * W                 # window (r, c) sends its gradient back to its 9 pixels
        used[r:r + 3, c:c + 3] += 1
print("how many windows covered each input pixel (= how many gradient terms it collects):\n", used)
print("\ndL/dX:\n", dX)
print("shortcut — 'full' convolution of delta (padding 2) with the kernel turned 180°:",
      np.allclose(dX, conv2d(delta, W[::-1, ::-1], padding=2)))
```

Corner pixels are in **1** window, the centre pixel in **9**: the centre collects 9 gradient terms. Now check every hand gradient two ways — the numerical slope (nudge one weight, watch the loss) and TensorFlow's automatic `GradientTape`.

```python
def loss_of(W_, b_):
    return 0.5 * np.sum((conv2d(X, W_) + b_ - target) ** 2)

eps = 1e-5
numeric = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        nudge = np.zeros((3, 3))
        nudge[i, j] = eps
        numeric[i, j] = (loss_of(W + nudge, b) - loss_of(W - nudge, b)) / (2 * eps)
print("numerical dL/dW:\n", numeric, "\nequal to ours?", np.allclose(numeric, dW, atol=1e-4))

X_tf, W_tf, b_tf = tf.Variable(X[None, :, :, None]), tf.Variable(W[:, :, None, None]), tf.Variable(np.float64(b))
with tf.GradientTape() as tape:
    out_tf = tf.nn.conv2d(X_tf, W_tf, strides=1, padding="VALID")[0, :, :, 0] + b_tf
    loss_tf = 0.5 * tf.reduce_sum((out_tf - target) ** 2)
g_X, g_W, g_b = tape.gradient(loss_tf, [X_tf, W_tf, b_tf])
print("\nTensorFlow agrees?  dW:", np.allclose(g_W.numpy()[:, :, 0, 0], dW),
      "| db:", np.isclose(g_b.numpy(), db), "| dX:", np.allclose(g_X.numpy()[0, :, :, 0], dX))

lr = 0.01
print(f"\none gradient-descent step (lr {lr}): loss {loss_of(W, b):g} ➜ {loss_of(W - lr * dW, b - lr * db):.2f}")
```

## Slightly more data: can gradient descent **find** the Sobel filter by itself?
Input: 500 MNIST digits. Target for each digit: what the Sobel x filter of Part 4 gives. Start from a **random** 3×3 kernel and use only the hand-written gradient above (averaged over all windows of all 500 digits).

```python
sobel = KERNELS["Sobel x"]
digits = x_train[:500]
windows = np.lib.stride_tricks.sliding_window_view(digits, (3, 3), axis=(1, 2))   # (500, 26, 26, 3, 3)
targets = np.einsum("nijkl,kl->nij", windows, sobel)

W_learn, b_learn, lr = rng.normal(0, 0.5, (3, 3)), 0.0, 1.0
print(f"{'step':>5} | {'loss':>8} | learned kernel (rounded)")
for step in range(301):
    delta = np.einsum("nijkl,kl->nij", windows, W_learn) + b_learn - targets
    if step in (0, 10, 50, 100, 300):
        print(f"{step:>5} | {0.5 * np.mean(delta ** 2):8.4f} | {np.round(W_learn, 1).tolist()}")
    W_learn -= lr * np.einsum("nijkl,nij->kl", windows, delta) / delta.size    # dL/dW: sum over every position
    b_learn -= lr * delta.mean()                                               # dL/db
print("\nSobel filter:", sobel.tolist(), f"| learned bias {b_learn:.3f}")
```

From random numbers, gradient descent recovers the Sobel filter — only by comparing outputs with targets. A real CNN does exactly this with many filters at once, except the "target" comes from the **label** (digit 7, cat, …) through the layers above, so the network invents whatever filters help.

# Part 6: colour images — 3 input channels
A colour image is H × W × **3** (red, green, blue). The kernel gets the same depth: 3 × 3 × **3**. Each channel is convolved with its own slice, then everything is **added up into one map**.

```python
rgb = rng.integers(0, 3, (4, 4, 3)).astype(float)                # tiny 4×4 colour image
kernel_rgb = rng.integers(-1, 2, (3, 3, 3)).astype(float)
per_channel = [conv2d(rgb[:, :, c], kernel_rgb[:, :, c]) for c in range(3)]
for c, name in enumerate(["red", "green", "blue"]):
    print(f"{name:>5} channel ⊛ its kernel slice:\n{per_channel[c]}")
print("\nsum of the three (one output map):\n", sum(per_channel))

def conv_layer(X, filters, bias, padding=0, stride=1):
    # X (H, W, C_in), filters (k, k, C_in, C_out), bias (C_out,) -> (H', W', C_out): any channels, any number of filters
    k = filters.shape[0]
    if padding:
        X = np.pad(X, ((padding, padding), (padding, padding), (0, 0)))
    windows = np.lib.stride_tricks.sliding_window_view(X, (k, k, X.shape[2]))[::stride, ::stride, 0]
    return np.einsum("ijklm,klmn->ijn", windows, filters) + bias

print("same with conv_layer, in one step?", np.allclose(conv_layer(rgb, kernel_rgb[..., None], np.zeros(1))[:, :, 0], sum(per_channel)))
```

On a real photo: a 3×3×3 filter with **+1/9 on red, −1/9 on green, 0 on blue** answers "is this patch more **red** than green?" — yes for the tan dog, no for the green grass. Edge filters see brightness; colour filters see colour.

```python
red_vs_green = np.zeros((3, 3, 3, 1))
red_vs_green[:, :, 0, 0], red_vs_green[:, :, 1, 0] = 1 / 9, -1 / 9               # +red −green, blue ignored
colour_map = np.maximum(0, conv_layer(colour["dog"], red_vs_green, np.zeros(1), padding=1))[:, :, 0]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].imshow(colour["dog"])
axes[0].set(title="photo (3 channels)", xticks=[], yticks=[])
axes[1].imshow(colour_map, cmap="gray_r")
axes[1].set(title="red − green filter (3×3×3) + ReLU: one map", xticks=[], yticks=[])
plt.tight_layout()
plt.show()
```

---
# Part 7: many filters ➜ many feature maps, and the weight count
A conv layer has many filters; each produces one **feature map**. Shapes: input (H, W, C_in) ➜ filters (k, k, C_in, C_out) ➜ output (H', W', C_out).
$$\text{weights} = k \cdot k \cdot C_{in} \cdot C_{out} + C_{out} \;(\text{biases})$$

```python
for c_in, c_out in [(1, 8), (3, 16), (16, 32), (64, 64)]:
    filters, bias = rng.normal(0, 0.1, (3, 3, c_in, c_out)), np.zeros(c_out)
    out = conv_layer(rng.random((28, 28, c_in)), filters, bias, padding=1)
    print(f"input (28, 28, {c_in:>2}) ➜ {c_out:>2} filters of 3×3×{c_in:<2} ➜ output {out.shape} | weights 3·3·{c_in}·{c_out} + {c_out} = {filters.size + bias.size:,}")

print(f"\n{'image':>16} | {'dense layer, 1000 neurons':>27} | {'conv layer, 64 filters 3×3':>27}")
print("-" * 78)
for name, (h, w, c) in [("MNIST 28×28×1", (28, 28, 1)), ("CIFAR 32×32×3", (32, 32, 3)), ("photo 224×224×3", (224, 224, 3))]:
    print(f"{name:>16} | {h * w * c * 1000 + 1000:>27,} | {3 * 3 * c * 64 + 64:>27,}")
```

The conv layer's weight count **does not depend on the image size**: the same small filters are reused at every position (weight sharing). The dense layer grows with every pixel.

# Part 8: pooling — shrink the map, keep the important part
2×2 windows, stride 2: **max pooling** keeps the strongest response, **average pooling** the mean. No weights to learn.

```python
def pool(X, size=2, mode="max"):
    windows = np.lib.stride_tricks.sliding_window_view(X, (size, size))[::size, ::size]
    return windows.max(axis=(2, 3)) if mode == "max" else windows.mean(axis=(2, 3))

fmap = np.array([[1, 3, 2, 0],
                 [4, 6, 1, 1],
                 [0, 2, 9, 5],
                 [1, 1, 3, 4]], dtype=float)
print("feature map:\n", fmap)
print("\ntop-left 2×2 window:", fmap[:2, :2].ravel(), "-> max", fmap[:2, :2].max(), "| average", fmap[:2, :2].mean())
print("\nmax pooling:\n", pool(fmap, mode="max"))
print("average pooling:\n", pool(fmap, mode="avg"))
```

On the photo: Sobel x ➜ ReLU ➜ max pool ➜ max pool. Each pooling keeps ¼ of the values, yet the outline survives.

```python
vertical_map = np.maximum(0, conv2d(gray["dog"], KERNELS["Sobel x"], padding=1))   # conv ➜ ReLU
pooled_1 = pool(vertical_map)
pooled_2 = pool(pooled_1)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.3))
for ax, img, title in [(axes[0], vertical_map, "Sobel x + ReLU"), (axes[1], pooled_1, "max pool"), (axes[2], pooled_2, "max pool again")]:
    ax.imshow(img, cmap="gray_r")
    ax.set(title=f"{title} {img.shape[0]}×{img.shape[1]}", xticks=[], yticks=[])
plt.tight_layout()
plt.show()
print(f"values: {vertical_map.size:,} ➜ {pooled_1.size:,} ➜ {pooled_2.size:,}")
```

---
# Part 9: pooling backward — where does the gradient go?
Pooling has **no weights**, but the gradient still has to pass through it to reach the conv layer below. Say the next layer sends back one gradient per pooled value:
- **max pooling**: only the pixel that **won** affected the output ➜ it gets the whole gradient, the other 3 get 0
- **average pooling**: all 4 pixels counted equally ➜ each gets ¼

```python
d_pooled = np.array([[1.0, 2.0],
                     [3.0, 4.0]])                                # gradient arriving from the next layer
d_max, d_avg = np.zeros_like(fmap), np.zeros_like(fmap)
for r in range(2):
    for c in range(2):
        window = fmap[2 * r:2 * r + 2, 2 * c:2 * c + 2]
        i, j = np.unravel_index(window.argmax(), window.shape)
        d_max[2 * r + i, 2 * c + j] = d_pooled[r, c]              # only the winner
        d_avg[2 * r:2 * r + 2, 2 * c:2 * c + 2] = d_pooled[r, c] / 4   # shared equally
print("feature map:\n", fmap, "\ngradient from above (2×2):\n", d_pooled)
print("\nmax pooling backward:\n", d_max, "\naverage pooling backward:\n", d_avg)

x_tf = tf.Variable(fmap[None, :, :, None])
with tf.GradientTape(persistent=True) as tape:
    loss_max = tf.reduce_sum(tf.nn.max_pool2d(x_tf, 2, 2, "VALID")[0, :, :, 0] * d_pooled)
    loss_avg = tf.reduce_sum(tf.nn.avg_pool2d(x_tf, 2, 2, "VALID")[0, :, :, 0] * d_pooled)
print("TensorFlow agrees?  max:", np.allclose(tape.gradient(loss_max, x_tf).numpy()[0, :, :, 0], d_max),
      "| average:", np.allclose(tape.gradient(loss_avg, x_tf).numpy()[0, :, :, 0], d_avg))
```

# Part 10: shifting the input — back to the problem of Part 1
Convolution: shift the image ➜ the feature map shifts by the same amount (**equivariance**). Pooling: small shifts barely change the result.

```python
edges = np.maximum(0, conv2d(digit, KERNELS["Sobel x"], padding=1))
shifted_edges = np.maximum(0, conv2d(np.roll(digit, 1, axis=1), KERNELS["Sobel x"], padding=1))
print("feature map of the shifted digit == shifted feature map?", np.allclose(shifted_edges[:, 2:-2], np.roll(edges, 1, axis=1)[:, 2:-2]))

def similarity(a, b):
    # 100% = identical pattern, 0% = nothing in common (cosine similarity)
    return (a.ravel() @ b.ravel()) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-12)

print("\naverage similarity to the unshifted digit, over 500 test digits")
print(f"{'shift':>5} | {'raw pixels':>10} | {'edge map':>8} | {'after 1 max pool':>16} | {'after 2 max pools':>17}")
print("-" * 70)
for shift in [1, 2, 3]:
    totals = np.zeros(4)
    for img in x_test[:500]:
        moved = np.roll(img, shift, axis=1)
        e = np.maximum(0, conv2d(img, KERNELS["Sobel x"], padding=1))
        e_moved = np.maximum(0, conv2d(moved, KERNELS["Sobel x"], padding=1))
        totals += [similarity(img, moved), similarity(e, e_moved), similarity(pool(e), pool(e_moved)),
                   similarity(pool(pool(e)), pool(pool(e_moved)))]
    t = 100 * totals / 500
    print(f"{shift:>5} | {t[0]:9.1f}% | {t[1]:7.1f}% | {t[2]:15.1f}% | {t[3]:16.1f}%")
```

Thin edge maps are very sensitive to shifts, but every pooling step makes the summary more similar again: after two poolings a digit moved 2 pixels still looks much like the original. This (plus stacking layers) is where a CNN's tolerance to small shifts comes from — the weakness of the MLP in Part 1.

# Part 11: receptive field — what one output "sees"
Feed a single bright pixel through stacked 3×3 convolutions and count how far it spreads.

```python
impulse = np.zeros((11, 11))
impulse[5, 5] = 1.0
x = impulse
for layers_count in range(1, 4):
    x = conv2d(x, np.ones((3, 3)), padding=1)
    rows = np.where(x.sum(axis=1) > 0)[0]
    print(f"after {layers_count} stacked 3×3 convolution(s): one pixel affects a {len(rows)}×{len(rows)} area")
```

Two 3×3 layers see 5×5, three see 7×7 — with fewer weights than one big 7×7 filter (3·9 = 27 vs 49). Deep CNNs stack small filters for this reason, and every pooling doubles what the next layer sees.

# Part 12: feature maps of a trained network (VGG16)
Our filters were **hand-made**. VGG16 (see the architectures notebook) **learned** its filters from 1.4 million ImageNet photos. Load it with its trained weights (about 58 MB, without the classifier on top) and look at the feature maps it makes for the dog photo — early, middle and deep in the network.

```python
vgg = keras.applications.VGG16(weights="imagenet", include_top=False)
layer_names = ["block1_conv2", "block3_conv3", "block5_conv3"]
probe = keras.Model(vgg.input, [vgg.get_layer(n).output for n in layer_names])

photo_224 = keras.utils.img_to_array(keras.utils.load_img(keras.utils.get_file(PHOTOS["dog"].split("/")[-1], PHOTOS["dog"]), target_size=(224, 224)))
vgg_maps = [m.numpy()[0] for m in probe(keras.applications.vgg16.preprocess_input(photo_224[None].copy()))]

fig, axes = plt.subplots(3, 8, figsize=(17, 7))
for row, (name, m) in enumerate(zip(layer_names, vgg_maps)):
    strongest = np.argsort(m.mean(axis=(0, 1)))[::-1][:8]          # the 8 maps that respond most to this photo
    for k, f in enumerate(strongest):
        axes[row, k].imshow(m[:, :, f], cmap="viridis")
        axes[row, k].set(xticks=[], yticks=[])
        axes[row, k].set_title(f"map {f}", fontsize=9)
    axes[row, 0].set_ylabel(f"{name}\n{m.shape[0]}×{m.shape[1]}×{m.shape[2]}", fontsize=10)
fig.suptitle("VGG16 feature maps of the same photo: 8 strongest maps of an early, a middle and a deep layer")
plt.tight_layout()
plt.show()
for name, m in zip(layer_names, vgg_maps):
    print(f"{name}: {m.shape[2]} feature maps of {m.shape[0]}×{m.shape[1]}")
```

- **Early layer** (224×224): maps still look like the photo — edges, outlines, colour areas, much like our hand-made filters of Part 4.
- **Middle layer** (56×56): outlines of the body, legs and paws, and the grass texture — combinations of edges.
- **Deep layer** (14×14): coarse blobs on **parts** — paws, legs, the back, the head — saying *where* something important is, no longer *what it looks like* pixel by pixel. Each value here sees a large part of the photo (receptive field, Part 11).

Do any of VGG16's **learned** first-layer maps look like our **hand-made** Sobel x filter? Compare our map with each of its 64 first-layer maps:

```python
first = keras.Model(vgg.input, vgg.get_layer("block1_conv1").output)
first_maps = first(keras.applications.vgg16.preprocess_input(photo_224[None].copy())).numpy()[0]   # (224, 224, 64)
ours = np.abs(conv2d(photo_224 @ np.array([0.299, 0.587, 0.114]) / 255.0, KERNELS["Sobel x"], padding=1))

match = np.array([np.corrcoef(ours.ravel(), first_maps[:, :, f].ravel())[0, 1] for f in range(64)])
best = np.argsort(match)[::-1][:3]
fig, axes = plt.subplots(1, 4, figsize=(15, 4))
axes[0].imshow(ours, cmap="gray_r")
axes[0].set(title="ours: |Sobel x| (hand-made)", xticks=[], yticks=[])
for ax, f in zip(axes[1:], best):
    ax.imshow(first_maps[:, :, f], cmap="gray_r")
    ax.set(title=f"VGG16 map {f}: correlation {match[f]:.2f}", xticks=[], yticks=[])
plt.tight_layout()
plt.show()
print(f"median correlation over all 64 learned maps: {np.median(match):.2f}")
```

Nobody told VGG16 to look for edges — it was only told the **labels** of 1.4 million photos. Its most similar first-layer maps match our hand-made edge detector with correlation about **0.47**, against a median of **0.20** over all 64 maps: **learning finds edge filters by itself**, like our 500-digit Sobel experiment in Part 5.

# Cheat sheet

| Operation | What it does | Shape / formula |
|---|---|---|
| convolution | slide a k×k filter, multiply, add up | output $= (n + 2p - k)/s + 1$ |
| padding p | zeros around the border | "same" for k = 3: p = 1 |
| stride s | jump s pixels each step | s = 2 halves the size |
| blur kernel | all positive, sum 1: average of the neighbourhood | box, Gaussian |
| edge kernel | positive vs negative side, sum 0: big where brightness changes | Sobel x / y, Laplacian |
| sharpen | identity − Laplacian: the photo plus its edges | sum 1 |
| weight sharing | the same k×k weights at every position | weights do not depend on image size |
| channels | filter depth = input channels, summed into one map | weights $k^2 C_{in} C_{out} + C_{out}$ |
| several filters | one feature map each (= one `Conv2D` layer) | output depth = number of filters |
| ReLU | keep positive responses | same shape |
| max / average pooling | strongest / mean of each window | 2×2, stride 2 halves the size, no weights |
| backward: dW | each weight was used at every position: add all the terms | $\partial L/\partial W = X \circledast \delta$ |
| backward: db | bias shared by all outputs | $\sum \delta$ |
| backward: dX | each pixel collects from every window that covered it | full convolution of $\delta$ with the kernel turned 180° |
| backward: pooling | max ➜ only the winner gets the gradient; average ➜ split equally | no weights to update |
| stacking | bigger receptive field | two 3×3 = 5×5, three 3×3 = 7×7 |

In this notebook **we** chose the filters (except the Sobel experiment and VGG16). Next notebook: our own `Conv2D` layer inside Keras, and a CNN that **learns** its filters on MNIST, tested on shifted digits.

```python
import pandas as pd
```

```python
visits_data = {
    "CustomerID": ["C1", "C2", 'C3', 'C4', 'C5', 'C6'],
    "Visit_Count": [5, 3, 8, 2, 7, 4]
}

purchase_data = {
    "CustomerID": ["C3", 'C4', 'C5', 'C7'],
    "Purchase_Amount":[200, 150, 300, 250]
}

visit_df = pd.DataFrame(visits_data)
purchase_df = pd.DataFrame(purchase_data)
```

```python
print("Visit data\n",visit_df,"\nPurchase data\n",purchase_df)
```

```python
common_df = pd.merge(visit_df, purchase_df, on="CustomerID", how="inner")
left_df = pd.merge(visit_df, purchase_df, on="CustomerID", how="left")
right_df = pd.merge(visit_df, purchase_df, on="CustomerID", how="right")
```

```python
print("customers who visited and purchased: ")
common_df
```

```python
print("customers who only visited and did not purchase:\n")
non_buyers = left_df[left_df["Purchase_Amount"].isna()]
non_buyers
```

```python
print("Customers who did not visit, but purchased:\n")
new_buyers = right_df[right_df["Visit_Count"].isna()]
new_buyers
```

```python
total_df = pd.merge(visit_df, purchase_df, on="CustomerID", how="outer")
```

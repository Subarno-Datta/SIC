```python
import numpy as np
```

```python
class Perceptron:

    def __init__(self, input_size: int, learning_rate: float):
        self.weights = np.zeros(input_size)
        self.bias = -0.8
        self.learning_rate = learning_rate

    def forward_pass(self, x: np.ndarray):
        return (1 if x @ self.weights + self.bias >= 0 else 0)

    def backpropagate(self, x: np.ndarray, error: int):
        self.weights += self.learning_rate*error*x
        self.bias += self.learning_rate*error

    def train(self, X: np.ndarray, Y: np.ndarray, epochs: int):
        for _ in range(epochs):
            for x, y_true in zip(X, Y):
                y_cap = self.forward_pass(x)
                error = y_true - y_cap
                self.backpropagate(x, error)
```

```python
X = np.array([
    [0,0],[0,1],[1,0],[1,1]
])
Y = np.array([0, 0, 0, 1])
```

```python
p = Perceptron(input_size=2, learning_rate=0.1)
```

```python
print("Before training:")
for x in X:
    print(f"Input: {x}, Prediction: {p.forward_pass(x)}")
```

```python
epochs = 3
p.train(X, Y, epochs=epochs)
```

```python
print("After training:")
for x in X:
    print(f"Input: {x}, Prediction: {p.forward_pass(x)}")
```

```python
print(f"\nFinal weights: {p.weights}")
print(f"\nFinal bias: {p.bias}")
```

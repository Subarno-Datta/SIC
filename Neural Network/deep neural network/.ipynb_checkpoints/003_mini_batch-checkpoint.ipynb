## Setup

```python
import io, time, urllib.request
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
np.set_printoptions(precision=4, suppress=True)
np.seterr(all="ignore")                                           # diverging runs overflow; we measure that instead

ACT = {
    "tanh": (np.tanh,                    lambda S: 1 - np.tanh(S) ** 2),
    "relu": (lambda S: np.maximum(0, S), lambda S: np.where(S > 0, 1.0, 0.0)),
}

def add_x0(X):
    return np.hstack([-np.ones((len(X), 1)), X])

def load_digits(name):
    url = "https://archive.ics.uci.edu/ml/machine-learning-databases/optdigits/" + name
    d = np.genfromtxt(io.StringIO(urllib.request.urlopen(url).read().decode()), delimiter=",").astype(int)
    return d[:, :64] / 16, d[:, 64]

Xd_tr, yd_tr = load_digits("optdigits.tra")
Xd_te, yd_te = load_digits("optdigits.tes")
Xd_tr0, Xd_te0, Yd_tr = add_x0(Xd_tr), add_x0(Xd_te), np.eye(10)[yd_tr]
print(f"digits: train {Xd_tr.shape}, test {Xd_te.shape}")

mnist = np.load(io.BytesIO(urllib.request.urlopen("https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz").read()))
Xm_tr, ym_tr = mnist["x_train"].reshape(-1, 784) / 255, mnist["y_train"]
Xm_te, ym_te = mnist["x_test"].reshape(-1, 784) / 255, mnist["y_test"]
Xm_tr0, Xm_te0, Ym_tr = add_x0(Xm_tr), add_x0(Xm_te), np.eye(10)[ym_tr]
print(f"MNIST : train {Xm_tr.shape}, test {Xm_te.shape}")
```

---
# Part 1: the problem — one sample at a time is slow
Our training so far (SGD): one forward + backward pass and one update **per sample**, in a Python loop.

```python
def per_sample_epoch(X0, Y, W1, W2, lr):
    # previous notebooks' style: one sample at a time (tanh hidden, softmax output)
    for i in rng.permutation(len(X0)):
        x, t = X0[i], Y[i]
        s1 = x @ W1
        h = np.r_[-1, np.tanh(s1)]
        s2 = h @ W2
        p = np.exp(s2 - s2.max()) / np.exp(s2 - s2.max()).sum()
        delta_out = t - p
        delta_hidden = (W2[1:] @ delta_out) * (1 - np.tanh(s1) ** 2)
        W2 = W2 + lr * np.outer(h, delta_out)
        W1 = W1 + lr * np.outer(x, delta_hidden)
    return W1, W2

start = time.time()
per_sample_epoch(Xm_tr0[:2000], Ym_tr[:2000], rng.normal(0, 0.036, (785, 128)), rng.normal(0, 0.09, (129, 10)), 0.05)
seconds = time.time() - start
print(f"MNIST, network 784-128-10, 2,000 images one at a time: {seconds:.1f} s")
print(f"All 60,000 images = 30 × that ≈ {seconds * 30:.0f} s for ONE epoch")
```

# Part 2: simple example — one slope per sample vs the average slope
Fit a line $\hat{y} = w x + b$ to 100 points. Loss $= \text{mean}((y - \hat{y})^2)$.

```python
x_line = rng.uniform(-1, 1, 100)
y_line = 2 * x_line + 1 + rng.normal(0, 0.3, 100)                # the true line is y = 2x + 1

w, b = rng.uniform(-3, 3, 2)
print(f"Start: w = {w:.3f}, b = {b:.3f}\n")
print("Slope of the loss for w, one sample at a time:  -2 × (y - ŷ) × x")
slopes = []
for i in range(4):
    error = y_line[i] - (w * x_line[i] + b)
    slopes.append(-2 * error * x_line[i])
    print(f"  sample {i}: x = {x_line[i]:+.3f}, y = {y_line[i]:+.3f}, error = {error:+.3f} -> slope = {slopes[-1]:+.3f}")
full_slope = np.mean(-2 * (y_line - (w * x_line + b)) * x_line)
print(f"\nAverage of these 4 slopes: {np.mean(slopes):+.3f} | average over all 100 samples (full batch): {full_slope:+.3f}")

print("\nHow far is a batch's average slope from the full-batch slope? (1,000 random batches each)")
for bs in [1, 4, 10, 25, 50]:
    gaps = []
    for _ in range(1000):
        i = rng.choice(100, bs, replace=False)
        gaps.append(abs(np.mean(-2 * (y_line[i] - (w * x_line[i] + b)) * x_line[i]) - full_slope))
    print(f"  batch of {bs:>2}: typical gap = {np.mean(gaps):.3f}")
```

One sample's slope is noisy. The bigger the batch, the closer its **average** slope gets to the full-batch slope.

```python
def line_path(batch_size, lr, epochs, w, b):
    path = [(w, b)]
    for _ in range(epochs):
        idx = rng.permutation(100)
        for s in range(0, 100, batch_size):
            i = idx[s:s + batch_size]
            error = y_line[i] - (w * x_line[i] + b)
            w = w - lr * np.mean(-2 * error * x_line[i])          # average slope over the batch
            b = b - lr * np.mean(-2 * error)
            path.append((w, b))
    return np.array(path)

wg, bg = np.meshgrid(np.linspace(-3.5, 4, 120), np.linspace(-3.5, 4, 120))
loss_map = ((y_line[None, None, :] - (wg[..., None] * x_line + bg[..., None])) ** 2).mean(axis=2)

fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))
for ax, (name, bs) in zip(axes, [("SGD (batch 1)", 1), ("mini-batch of 10", 10), ("full batch (100)", 100)]):
    path = line_path(bs, 0.05, 5, w, b)
    ax.contour(wg, bg, loss_map, levels=30, cmap="Greys")
    ax.plot(path[:, 0], path[:, 1], "o-", ms=2, lw=1, color="crimson")
    ax.plot(2, 1, "g*", ms=15)
    ax.set(xlabel="w", ylabel="b", title=f"{name}: {len(path) - 1} updates in 5 epochs")
plt.tight_layout()
plt.show()
```

SGD: many small, zig-zag steps. Full batch: smooth but only one step per epoch. Mini-batch: in between.

# Part 3: the same functions on a whole batch at once

| Function | One sample (before) | A batch of $m$ samples (now) |
|---|---|---|
| `forward_pass` | $s = h \cdot W$ | $S = H\,W$ (one row per sample) |
| `compute_error` | $y - p$ | $Y - P$ (one row per sample) |
| `backward_pass` | $dW = -h \otimes \delta$ | $dW = -H^\top \Delta \,/\, m$ (**average** over the batch) |

```python
def new_net(sizes, act="tanh"):
    k = 1 if act == "tanh" else 2                                 # Xavier for tanh, He for relu
    return {"act": act, "W": [rng.normal(0, np.sqrt(k / a), (a + 1, b)) for a, b in zip(sizes[:-1], sizes[1:])]}

def forward_pass(X0, net):
    f = ACT[net["act"]][0]
    H, cache = X0, []
    for W in net["W"][:-1]:
        S = H @ W
        cache.append((H, S))
        H = add_x0(f(S))
    cache.append((H, None))
    S = H @ net["W"][-1]
    E = np.exp(S - S.max(axis=1, keepdims=True))
    return E / E.sum(axis=1, keepdims=True), cache

def compute_error(Y, P):
    return Y - P

def backward_pass(net, error, cache, lr):
    df = ACT[net["act"]][1]
    delta, new_W = error, list(net["W"])
    for l in range(len(net["W"]) - 1, -1, -1):
        dW = -cache[l][0].T @ delta / len(delta)                  # average slope over the batch
        if l > 0:
            delta = (delta @ net["W"][l][1:].T) * df(cache[l - 1][1])
        new_W[l] = net["W"][l] - lr * dW
    return {"act": net["act"], "W": new_W}

def accuracy(net, X0, y):
    return 100 * (forward_pass(X0, net)[0].argmax(axis=1) == y).mean()

def train(X0, Y, net, lr, epochs, batch_size, X_test=None, y_test=None):
    history = []                                                  # (seconds so far, test accuracy) after each epoch
    start = time.time()
    for _ in range(epochs):
        idx = rng.permutation(len(X0))
        for s in range(0, len(X0), batch_size):
            batch = idx[s:s + batch_size]
            P, cache = forward_pass(X0[batch], net)
            net = backward_pass(net, compute_error(Y[batch], P), cache, lr)
        if X_test is not None:
            history.append((time.time() - start, accuracy(net, X_test, y_test)))
    return net, history
```

## Shape walk for a batch of 4

```python
net = new_net([64, 64, 64, 10])
P, cache = forward_pass(Xd_tr0[:4], net)
for l, (H, S) in enumerate(cache[:-1]):
    print(f"layer {l + 1}: H {H.shape} @ W {net['W'][l].shape} -> S {S.shape}")
print(f"output : H {cache[-1][0].shape} @ W {net['W'][-1].shape} -> P {P.shape} (4 samples × 10 probabilities)")
```

## Check: batch update = average of the 4 one-sample updates

```python
batch_update = backward_pass(net, compute_error(Yd_tr[:4], P), cache, lr=1.0)
one_by_one = []
for i in range(4):
    Pi, cache_i = forward_pass(Xd_tr0[i:i + 1], net)
    one_by_one.append(net["W"][0] - backward_pass(net, compute_error(Yd_tr[i:i + 1], Pi), cache_i, lr=1.0)["W"][0])
dW_batch = net["W"][0] - batch_update["W"][0]
print("dW (first layer, first 3 values) from the batch            :", dW_batch.ravel()[:3])
print("average of the 4 one-sample dW (first layer, first 3 values):", np.mean(one_by_one, axis=0).ravel()[:3])
print("Equal?", np.allclose(dW_batch, np.mean(one_by_one, axis=0)))
```

## Speed: the same 1,000 samples, loop vs one matrix

```python
start = time.time()
for i in range(1000):
    forward_pass(Xd_tr0[i:i + 1], net)
loop_time = time.time() - start
start = time.time()
forward_pass(Xd_tr0[:1000], net)
matrix_time = time.time() - start
print(f"loop over 1,000 samples: {loop_time * 1000:.1f} ms | one batch of 1,000: {matrix_time * 1000:.1f} ms | {loop_time / matrix_time:.0f}× faster")
```

# Part 4: digits — batch size 1 vs 8 vs 32 vs 128 vs full
Network `[64, 64, 64, 10]`, tanh, Xavier. Each batch size uses its best learning rate (see Experiment 1).

```python
setups = [(1, 0.05, 10), (8, 0.2, 10), (32, 0.2, 20), (128, 0.2, 40), (len(Xd_tr0), 0.2, 300)]   # (batch size, η, epochs)
fig, axes = plt.subplots(1, 2, figsize=(15, 4))
print(f"{'batch':>6} | {'η':>4} | {'updates/epoch':>13} | {'s/epoch':>7} | {'epochs':>6} | {'test acc':>8}")
print("-" * 60)
for bs, lr, epochs in setups:
    _, hist = train(Xd_tr0, Yd_tr, new_net([64, 64, 64, 10]), lr, epochs, bs, Xd_te0, yd_te)
    hist = np.array(hist)
    print(f"{bs:>6} | {lr:>4} | {int(np.ceil(len(Xd_tr0) / bs)):>13} | {hist[-1, 0] / epochs:7.3f} | {epochs:>6} | {hist[-1, 1]:7.1f}%")
    axes[0].plot(range(1, epochs + 1), hist[:, 1], label=f"batch {bs}")
    axes[1].plot(hist[:, 0], hist[:, 1], label=f"batch {bs}")
axes[0].set(xlabel="epoch", ylabel="test accuracy (%)", xscale="log", title="Per epoch: small batches learn more (more updates)")
axes[1].set(xlabel="seconds (log scale)", ylabel="test accuracy (%)", xscale="log", title="Per second: mini-batches win")
axes[0].legend()
axes[1].legend()
plt.show()
```

# Part 5: real use case — MNIST (60,000 handwritten digits, 28×28)

```python
fig, axes = plt.subplots(1, 10, figsize=(15, 1.8))
for d, ax in enumerate(axes):
    ax.imshow(Xm_tr[ym_tr == d][0].reshape(28, 28), cmap="gray_r")
    ax.set(title=str(d), xticks=[], yticks=[])
plt.show()
```

```python
mnist_net, mnist_hist = train(Xm_tr0, Ym_tr, new_net([784, 128, 64, 10], "relu"), 0.1, 5, 64, Xm_te0, ym_te)
for epoch, (seconds, acc) in enumerate(mnist_hist, 1):
    print(f"Epoch {epoch} | {seconds:5.1f} s so far | test accuracy {acc:.2f}%")
```

```python
P = forward_pass(Xm_te0, mnist_net)[0]
pred = P.argmax(axis=1)
cm = np.zeros((10, 10), dtype=int)
for t, p in zip(ym_te, pred):
    cm[t, p] += 1

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.imshow(cm, cmap="Blues")
for i in range(10):
    for j in range(10):
        ax.text(j, i, cm[i, j], ha="center", va="center", fontsize=8, color="white" if cm[i, j] > cm.max() / 2 else "black")
ax.set(xticks=range(10), yticks=range(10), xlabel="predicted", ylabel="actual", title=f"MNIST test: {100 * (pred == ym_te).mean():.2f}% correct")
plt.show()

wrong = np.where(pred != ym_te)[0][:10]
fig, axes = plt.subplots(1, 10, figsize=(15, 2))
for ax, i in zip(axes, wrong):
    ax.imshow(Xm_te[i].reshape(28, 28), cmap="gray_r")
    ax.set(title=f"true {ym_te[i]}, said {pred[i]}", xticks=[], yticks=[])
plt.show()
```

# Part 6: Experiments (digits, `[64, 64, 64, 10]`, 10 epochs)
## 1. Batch size × learning rate

```python
batch_sizes = [1, 8, 32, 128, len(Xd_tr0)]
lrs = [0.01, 0.05, 0.2, 1.0]
grid = np.zeros((len(batch_sizes), len(lrs)))
for r, bs in enumerate(batch_sizes):
    for c, lr in enumerate(lrs):
        grid[r, c] = accuracy(train(Xd_tr0, Yd_tr, new_net([64, 64, 64, 10]), lr, 10, bs)[0], Xd_te0, yd_te)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.imshow(grid, cmap="Greens", vmin=0, vmax=100)
for r in range(len(batch_sizes)):
    for c in range(len(lrs)):
        ax.text(c, r, f"{grid[r, c]:.0f}%", ha="center", va="center")
ax.set(xticks=range(len(lrs)), xticklabels=lrs, yticks=range(len(batch_sizes)), yticklabels=batch_sizes,
       xlabel="learning rate η", ylabel="batch size", title="Test accuracy after 10 epochs")
plt.show()
```

Small batches break at large η (each step follows one noisy sample). Bigger batches average out the noise, so they can take bigger steps. Full batch makes only 10 updates in 10 epochs.

## 2. How noisy is each update? Training loss after each of the first 150 updates

```python
def loss_after_each_update(batch_size, lr, updates=150):
    net, losses = new_net([64, 64, 64, 10]), []
    idx = rng.permutation(len(Xd_tr0))
    for u in range(updates):
        batch = idx[(u * batch_size) % len(idx):][:batch_size]
        P, cache = forward_pass(Xd_tr0[batch], net)
        net = backward_pass(net, compute_error(Yd_tr[batch], P), cache, lr)
        P_all = forward_pass(Xd_tr0, net)[0]
        losses.append(-np.mean(np.log(np.sum(Yd_tr * P_all, axis=1) + 1e-12)))
    return losses

fig, ax = plt.subplots(figsize=(9, 3.5))
for bs, lr in [(1, 0.05), (32, 0.2), (512, 0.2)]:
    ax.plot(loss_after_each_update(bs, lr), label=f"batch {bs}, η = {lr}")
ax.set(xlabel="update", ylabel="training loss (all samples)", title="Bigger batch = smoother steps")
ax.legend()
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
print(f"Tensorflow version: {tf.__version__}")
print(f"Keras version: {keras.__version__}")
```

```python
print("Generating data")
x_train = np.random.random((1000000, 50)).astype('float32')
y_train = np.random.randint(2, size=(1000000, 1)).astype('float32')

print(f"X_train:\n{x_train}\n\n")
print(f"Y_train:\n{y_train}")
```

```python
print("BUILDING A SIMPLE NEURAL NETWORK")
model = keras.Sequential([
    layers.Input(shape=(50,)),
    layers.Dense(128, activation='leaky_relu'),
    layers.Dense(512, activation='gelu'),
    layers.Dense(1024, activation='leaky_relu'),
    layers.Dense(2048, activation='gelu'),
    layers.Dense(4096, activation='leaky_relu'),
    layers.Dense(2048, activation='leaky_relu'),
    layers.Dense(1024, activation='gelu'),
    layers.Dense(512, activation='leaky_relu'),
    layers.Dense(256, activation='gelu'),
    layers.Dense(128, activation='leaky_relu'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
keras.

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
print("TESTING THE MODEL TRAINING (fit)")
model.fit(x_train, y_train, epochs=5, batch_size=1024, verbose=1)
```

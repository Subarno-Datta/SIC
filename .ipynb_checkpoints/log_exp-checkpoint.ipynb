```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = 'https://raw.githubusercontent.com/selva86/datasets/master/Advertising.csv'
data = pd.read_csv(url)
```

```python
data
```

```python
x = data['TV']
y = data['sales']
```

```python
coeff = np.polyfit(x, y, 2)
poly = np.poly1d(coeff)
```

```python
x_new = np.linspace(x.min(), x.max(), 100)
y_new = poly(x_new)
```

```python
plt.scatter(x, y)
plt.plot(x_new, y_new)
plt.title('Polynomial Regression. TV Budget vs sales')
plt.xlabel('TV Advertising Budget')
plt.ylabel('Sales')
plt.grid()
plt.show()
```

```python
def grow(x):
    return 2**x

def decay(x):
    return (0.5)**x

a = np.linspace(-5, 5, 21)
plt.plot(a, grow(a), label='f(x) = 2$^x$', color='black')
plt.plot(grow(a), a, label='f(x) = $log(x)/log(2)$', color='purple')
plt.plot(np.linspace(-5, 30, 61), np.linspace(-5, 30, 61), color='yellow')
plt.grid()
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
days = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
cases = np.array([5, 9, 20, 40, 85, 170])

log_cases = np.log(cases)
```

```python
model = LinearRegression()
model.fit(days, log_cases)
```

```python
m = np.array(sorted(np.random.randint(0, 15, size=35)))
predicted_log = model.predict(m.reshape(-1, 1))
predicted_cases = np.exp(predicted_log)
```

```python
plt.scatter(days, cases, label='Actual Data')
plt.plot(m, predicted_cases, label='Exponential Fit')
plt.title('Exponential Growth - Infection Spread')
plt.xlabel("Days")
plt.ylabel("Cases")
plt.legend()
plt.grid()
plt.show()
```

```python
data = np.array([1, 5, 10, 50, 100, 500, 1000])
log_data = np.log(data)
```

```python
log_data
```

```python
plt.plot(data, log_data, label='ln(x) compression')
plt.xlabel('orignal data')
plt.ylabel('log(data)')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.plot(log_data, data, label='exp(x) expansion')
plt.xlabel('log(data)')
plt.ylabel('original data')
plt.grid()
plt.legend()
plt.show()
```

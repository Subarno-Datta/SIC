```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 100)
y_linear = x
```

```python
y_sigmoid = 1/(1 + np.exp(-x))
```

```python
plt.figure(figsize=(5000,25))
```

```python
hours = np.array([1, 3, 5, 7])
z = 1.2 * hours - 4
sigmoid = 1/(1 + np.exp(-z))
zh = 1.2 * 3.5 -4
sigmoid_g = 1/(1 + np.exp(-zh))

plt.subplot(1, 3, 1)
plt.plot(x, y_linear)
plt.title('Before Transformation')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()

plt.subplot(1, 3, 2)
plt.plot(x, y_sigmoid)
plt.title('sigmoid')
plt.xlabel('x')
plt.ylabel('Probability')
plt.grid()

plt.subplot(1, 3, 3)
plt.plot(hours, sigmoid)
plt.title('pass probability')
plt.xlabel('Hours studied')
plt.ylabel('Probability of passing')
plt.grid()

plt.tight_layout()
```

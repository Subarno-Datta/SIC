```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = lambda x: 3 + 4*x + x*x
g = lambda x: x**3 - 7*x*x + 4*x + 2

dob = lambda d: 2026 - d
```

```python
x = np.linspace(-5, 5, 11)
y1 = f(x)
y2 = g(x)

dobs = np.random.randint(1965, 2024, (2024-1965+1, 1))

age = dob(dobs)
```

```python
plt.ylim(0, 100)
# plt.plot(x, y1, label='$f(x) = x^2$', color='red')
# plt.plot(x, y2, label='$g(x) = x^3$', color='green')
plt.plot(dobs, age, label='age = current year - date of birth', color='green')
# plt.title('Plot of $f(x) = x^2$ and $g(x) = x^3$')
plt.ylabel('AGE')
plt.xlabel('YEAR OF BIRTH')
plt.grid(True)
plt.legend()
plt.show()
# plt.plot(x, y2, label='$g(x) = x^3', color='green')
```

```python
def n_term(a, n, d):
    return a + (n-1)*d
```

```python
def sum_seq(a, d, n):
    return int(n*(2*a + d*(n-1))*0.5)
```

```python
sum_seq(12, 4, 25)
```

```python
def is_gp(li):
    if len(li)==0 or len(li)==1:
        return True
    p = li[1]/li[0]
    for i in range(1, len(li)-1):
        if (li[i+1]/li[i] == p):
            continue
        return False
    return True
    

def is_ap(li):
    if len(li)==0 or len(li)==1:
        return True
    d = li[0] - li[1]
    for i in range(1, len(li)-1):
        if (li[i] - li[i+1] == d):
            continue
        return False
    return True

is_seq = lambda li: is_ap(li) or is_gp(li)
```

```python
is_seq([5, 6, 7, 8])
```

```python
is_seq([5, 6, 8, 9, 11])
```

```python
is_seq([2, 4, 8, 16, 32])
```

```python
is_seq([2, 4, 6, 12, 14, 28])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
time_index = pd.date_range(start="2026-03-25 00:00:00", periods=10, freq='h')
temperature = np.random.randint(20, 35, size=10)
```

```python
df = pd.DataFrame({
    'Timestamp': time_index,
    'Temperature': temperature
})
```

```python
df.set_index('Timestamp', inplace=True)
```

```python
print('Time series data: ')
df
```

```python
print(f'Mean temperature: {df["Temperature"].mean()}')
```

```python
print(f'Max temperature: {df["Temperature"].max()}')
```

```python
print(f'Min temperature: {df["Temperature"].min()}')
```

```python
plt.plot(df.index, df['Temperature'], label='Temperature')
plt.legend()
plt.show()
```

```python
marks = np.random.randint(0, 100, size=70)
scaled = (marks - marks.min()) / (marks.max() - marks.min())
```

```python
bins = np.linspace(marks.min(), marks.max(), 10)
```

```python
bins
```

```python
labels = [str(i) for i in range(2, 11)]
grades = pd.cut(marks, bins=bins, labels=labels, include_lowest=True)
dataframe = pd.DataFrame({
    'marks':marks,
    'scaled': scaled,
    'grades': grades
})
```

```python
a = 1000
r = 1.2

details = {}
for i in range(100):
    details[i] = a * r**i

growth = pd.DataFrame({
    'Day': details.keys(),
    'Value': details.values()
})
plt.plot(growth.index, growth['Value'], label='Population')
plt.xlabel('$Days$')
plt.ylabel('Population (x$10^1$$^0$)')
plt.legend()
plt.show()
```

```python
ini = 100
decay = 0.8

decay_details = {}

for i in range(100):
    decay_details[i] = ini * decay**i

decay_df = pd.DataFrame({
    'Hour': decay_details.keys(),
    'Battery Percentage': decay_details.values()
})

plt.plot(decay_df.index, decay_df['Battery Percentage'], label='Battery Percentage')
plt.xlabel('Hours')
plt.ylabel('Battery Percentage')
plt.legend()
plt.show()
```

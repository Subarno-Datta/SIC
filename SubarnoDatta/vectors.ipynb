L1 NORM - for v = [v1, v2, v3, v4, ..., vn], then L1 Norm = |v1|+|v2|...+|vn|

L2 NORM - L2 Norm --> if S = v1^2 + v2^2 + .... + vn^2, then L2 Norm = sqrt(S), same as |v|, or ||v||

L[inf] NORM ---> the L<inf> Norm is the Max value of |vi| for i = 1 to n.. , for example x = [-2, 76, -101], then L[inf] norm of x is 101

```python
products = np.array([200, 150, 300, 250])
quantity = np.array([5, 2, 3, 4])
print(int(products.dot(quantity)))
```

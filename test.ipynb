```python
import pandas as pd
import numpy as np
```

```python
def load_boston():
    data_url = "http://lib.stat.cmu.edu/datasets/boston"
    raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)
    data = np.hstack([raw_df.values[::2,:], raw_df.values[1::2, :2]])
    target = raw_df.values[1::2, 2]
    return {'data': data, 'target': target}
```

```python
boston = load_boston()
X = boston['data']
y = boston['target']
A = X[:4, [0,4,5,6]] #CRIM, NOX, RM, AGE
b = y[:4]
```

```python
x = np.linalg.solve(A, b)
```

```python
print(x)
```

```python
np.linalg.inv(A.T @ A) @ A.T @ b
```

```python
X_w = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
x, y, a, b, k = sp.symbols('x y a b k')
```

```python
fx = x**3 - 1
gx = sp.log(sp.log(x**2 - 3*k))
hx = sp.exp(a*x**b)
```

```python
sp.diff(fx, x)
```

```python
sp.diff(gx, x)
```

```python
sp.diff(hx, x)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import random
```

```python
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
```

```python
y = np.array([3, 4, 5, 6, 7])
```

```python
model = LinearRegression()
```

```python
model.fit(x, y)
```

```python
predicted_salary = model.predict([[6]])
```

```python
predicted_salary
```

```python
plt.scatter(x, y, label='actual data')
plt.plot(x, model.predict(x), label='Regression line')
plt.xlabel('Experience (years)')
plt.ylabel('Salary')
plt.legend()
plt.grid()
plt.show()
```

```python
a = np.array([random.randint(0, 100) for i in range(100)]).reshape(-1, 1)
b = np.array([random.randint(-8, 108) for i in range(100)])
plt.scatter(a, b)
model.fit(x, y)
plt.title('Scattered Points')
plt.plot(a, model.predict(a), label='Regression Line')
plt.grid()
plt.legend()
plt.show()
```

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
```

```python
n=10
p=0.5
st.binom.pmf(5, n, p)
```

```python
st.binom.ppf(0.3, n, p)
```

```python
st.binom.cdf(5, n, p)
```

```python
st.binom.cdf(7, n, p) - st.binom.cdf(2, n, p) # P(3 <= x <= 7)
```

```python
x = np.arange(0,11)
```

```python
plt.plot(x, st.binom.pmf(x, n, p), color='red')
```

```python
lamb = 2
st.poisson.ppf(0.5, lamb)
```

```python
st.poisson.cdf(5, lamb)
```

```python
st.poisson.cdf(7, lamb) - st.poisson.cdf(2, lamb)
```

```python
plt.scatter(x, st.poisson.pmf(x, lamb), color='green')
```

```python
print("UNIFORM DISTRIBUTION")
uniform_data = np.random.uniform(1, 100, 20)
print("Marks: ", np.round(uniform_data, 2))

normal_data = np.random.normal(70, 10, 20)
print("\nNORMAL DISTRIBUTION")
print("\n20 students marks:")
print(np.round(normal_data, 2))
```

```python
print("Mean marks: ", round(np.mean(normal_data), 2))
print("Standard Deviation: ", round(np.std(normal_data), 2))
print("Highest mark: ", round(np.max(normal_data), 2))
print("Lowest mark: ", round(np.min(normal_data), 2))
```

```python
low = 50
high = 80

print(len([x for x in uniform_data if low <= x <= high])/20)
print(len([x for x in normal_data if low <= x <= high])/20)
```

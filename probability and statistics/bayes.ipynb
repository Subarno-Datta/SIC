```python
repays = [0.5, 0.7, 0.9, 0.4, 0.8, 0.48]
defaults = [1-x for x in repays]

for i in range(len(repays)):
    p_repay = repays[i]
    p_default = defaults[i]

    print(f"\nHistorical repayment probability for person {i+1}: {p_repay*100}%")

    p_goodcredit = p_repay*p_goodCredit_repay + p_default*p_goodCredit_default
    p_repay_goodCredit = p_repay*p_goodCredit_repay/p_goodcredit
    print(f"Probability of repayment of loan given good credit score: {round(p_repay_goodCredit, 4)*100}%")
    if p_repay_goodCredit*100 > 75:
        print(f"Loan approved for person {i+1}")
    else:
        print(f"Loan not approved for person {i+1}")
    
```

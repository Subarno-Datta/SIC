```python
import numpy as np
import scipy.stats as st
```

```python
x = np.random.randint(10, 30, 40)
```

```python
x
```

```python
n = len(x)
```

```python
n
```

```python
x_mean = x.mean()
SEM = x.std(ddof=1)/np.sqrt(n*1.0)
```

```python
(x_mean - 1.645 * SEM, x_mean + 1.645*SEM)
```

```python
(x_mean - st.norm.ppf(0.95,loc=0,scale=1)*SEM, x_mean + st.norm.ppf(0.95,loc=0,scale=1)*SEM)
```

```python
st.norm.interval(0.9, loc=x_mean, scale=SEM)
```

```python
(x_mean - st.t.ppf(0.995,df=n-1)*SEM, x_mean + st.t.ppf(0.995,df=n-1)*SEM)
```

```python
st.t.interval(0.9995, df=n-1, loc=x_mean, scale=SEM)
```

```python
st.t.interval(0.9999, df=n-1, loc=x_mean, scale=SEM)
```

## INTERVAL ESTIMATION OF PROPORTION

```python
import numpy as np
import scipy.stats as st
```

1. There are two candidates for the election: A and B. The candidate A wants to survey his approval rating. Out of 100 surveyed, 55 answered positively. Can A be sure of this election?
   - Assume that A gets elected with 50% or more
   - Use 95% confidence interval

```python
n_yes = 55
n_total = 100
p_mean = n_yes/n_total
SE = np.sqrt(p_mean*(1-p_mean)/n_total)
SE2 = np.sqrt(p_mean*(1-p_mean)/(n_total*10))
```

```python
st.norm.interval(0.95, loc=p_mean, scale=SE)
```

=> *The confidence interval has the lower bound below 50%. So A cannot be sure with the confidence level of 95%*

```python
print(st.norm.interval(0.99, loc=p_mean, scale=SE))
print(st.norm.interval(0.90, loc=p_mean, scale=SE))
print(st.norm.interval(0.70, loc=p_mean, scale=SE))
```

```python
print(st.norm.interval(0.95, loc=p_mean, scale=SE2))
```

=> *The confidence interval has the lower bound above 50%. So A can be sure with the confidence level of 95%*

## INTERVAL ESTIMATION OF CORRELATION

```python
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df
```

```python
x = df['petal_length']
y = df['sepal_length']
```

```python
np.round(st.pearsonr(x,y),3)
```

```python
x.corr(y)
```

```python
np.round(df.drop(columns=['species']).corr(),3)
```

```python
np.round(st.spearmanr(x,y),3)
```

```python
np.round(st.kendalltau(x,y),3)
```

```python
n = len(x)
r = x.corr(y)
z = np.arctanh(r)
std_error_z = 1/np.sqrt(n-3)
```

```python
{'low': np.tanh(z-st.norm.ppf(0.975)*std_error_z), 'high': np.tanh(z+st.norm.ppf(0.975)*std_error_z)}
```

```python
{'low': np.tanh(z-st.norm.ppf(0.995)*std_error_z), 'high': np.tanh(z+st.norm.ppf(0.995)*std_error_z)}
```

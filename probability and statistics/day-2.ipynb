```python
import random
```

```python
even = [2, 4, 6]
prime = [2, 3, 5]
evenc, oddc, primec = 0,0,0
results = []
```

```python
for i in range(20):
    x = random.randint(1, 6)
    results.append(x)
    if x in even:
        evenc += 1
    else:
        oddc += 1
    if x in prime:
        primec += 1

print(f'Results: {results}')
print(f'Even outcomes:\t{evenc}\t,\tprobability = {evenc/20}')
print(f'Odd outcomes:\t{oddc}\t,\tprobability = {oddc/20}')
print(f'Prime outcomes:\t{primec}\t,\tprobability = {primec/20}')

results = []
evenc,oddc,primec = 0,0,0
```

```python
total_students = 30
instagram_users = 24
instagram_and_educational = 18

P_instagram = instagram_users/total_students
P_instagram_and_educational = instagram_and_educational/total_students
P_educational_given_instagram = instagram_and_educational/instagram_users
```

```python
print(f'Total students: {total_students}')
print(f'Students using instagram: {instagram_users}')
print(f'Students using instagram and watching educational videos: {instagram_and_educational}')

print("\nPROBABILITY CALCULATIONS")
print(f'P(Instagram) = {round(P_instagram, 2)}')
print(f'P(Instagram and Educational Vidoes) = {round(P_instagram_and_educational, 2)}')
print(f'P(Educational Videos | Instagram) = {round(P_educational_given_instagram, 2)}')

print("\nINTERPRETATION")
print(f'{P_educational_given_instagram*100:.0f}% of Instagram users watch educational content')
```

```python
import numpy as np
from scipy.stats import binom, poisson
```

```python
p_spam = 0.2
email = np.random.binomial(1, p_spam)
print("Probability of Spam = ", p_spam)
```

```python
if email==1:
    print('Result: the email is SPAM')
else:
    print('Result: the email is not SPAM')

print("Interpretation:")
print("Bernoulli Distribution models a single email")
print("Possible outcomes:")
print("1->Spam")
print("2->Not Spam")

n = 15
x = 4
binomial_probability = binom.pmf(x, n, p_spam)
print("Total emails (n) = ", n)
print("Probability of spam (p) =", p_spam)
print("Required spam emails (x) =", x)

print("\nProbability of exactly 4 spam emails: ")
print("P(X=4) =", round(binomial_probability, 6))
```

```python
lam = 3
x = 5
poisson_probability = poisson.pmf(x, lam)
print("average spam emails per minute (lambda) = ",lam)
print("required spam emails (x) =", x)
print("P(x=5) =", round(poisson_probability, 6))

print("\nInterpretation:")
print("Poisson distribution models even occuring")
```

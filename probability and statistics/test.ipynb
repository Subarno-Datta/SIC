```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
df = pd.DataFrame(data.data, columns=[data.feature_names])
df = df.assign(target=pd.Series(1-data.target))
```

```python
df
```

```python
df.corr()
```

```python
corre = df.corr()['target'].drop('target')
```

```python
corre
```

```python
strongest_feature = corre['target'].abs().idxmax()
actual_value = corre.loc[strongest_feature, 'target']
```

```python
print(f'top correlation: {strongest_feature}, value: {actual_value}')
```

```python
import numpy as np

a = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

b = np.linalg.inv(a @ a.T)

c = np.array([[1, 2], [3, 4]])
d = np.linalg.inv(c)

print(a)
print(a.T)
print(b)
print(a.T @ a)
print(c)
print(d)
```

```python
A = np.array([
    [1, 2, 6],
    [1, 4, 5],
    [1, 6, 7]
])

B = np.array([50, 60, 80])

r_cap = np.linalg.inv(A.T @ A) @ A.T @ B

print(r_cap)
```

```python
matrix = np.array([
    [0, 0, 3],
    [0, 0, 0],
    [1, 0, 0]
])

zero_count = np.sum(matrix==0)
total_elements = matrix.size

if zero_count > total_elements/2:
    print('Sparse matrix')
else:
    print('Not a sparse matrix')
```

```python
rows, cols = matrix.shape
sparse_representation = []

for i in range(rows):
    for j in range(cols):
        if matrix[i][j]!=0:
            sparse_representation.append((i, j, int(matrix[i][j])))

print(sparse_representation)
```

```python
from scipy.sparse import csr_matrix

A_dense = (csr_matrix(sparse_representation)).toarray()
```

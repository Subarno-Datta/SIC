```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
coffee = pd.read_csv('data_coffeeshop.csv')
```

```python
coffee.head()
```

```python
freq = coffee['Year of Start'].dropna().astype(int).value_counts().sort_index()
```

```python
plt.plot(freq.index, freq.values, marker='o', color='b', linestyle='-')
plt.xlabel('Year of Start')
plt.ylabel('Frequency')
plt.xlim([1997, 2014])
```

```python
IN = coffee[coffee['Current State']=='In']
OUT = coffee[coffee['Current State']=='Out']
```

```python
IN = IN['Year of Start'].dropna().astype(int).value_counts().sort_index()
OUT = OUT['Year of Start'].dropna().astype(int).value_counts().sort_index()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(IN.index, IN.values, marker='o', color='blue', linestyle='-', label='Business In')
plt.plot(OUT.index, OUT.values, marker='o', color='red', linestyle='-', label='Business Out')
plt.xlabel('Year of Start')
plt.ylabel('Frequency')
plt.legend()
plt.xlim([1997, 2014])
```

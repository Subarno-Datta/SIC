```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.random.randn(10000)
plt.hist(x, bins=100, color='blue', density=False, alpha=0.4)
plt.title('HISTOGRAM')
plt.show()
```

```python
x = np.random.randn(10000)*10 + 3
y = np.random.randn(10000)*10 + 5
z = np.random.randn(10000)*10 + 1
plt.boxplot([x,y,z],vert=False)
plt.title('BOX PLOTS')
plt.show()
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
plt.plot(x,y,marker='o',markerfacecolor='yellow', markeredgewidth=2, markeredgecolor='purple',linewidth=2,linestyle='none',markersize=15)
plt.xlabel('X')
plt.ylabel('$y = sin(x)$')
# plt.xlim([0,5])
# plt.ylim([-2, 2])
plt.title('LINE PLOT')
plt.show()
```

```python
z = np.cos(x)
```

```python
fig0 = plt.figure()
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x,y,color='purple',linestyle='-')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_xlim([0,6])
axes0.set_ylim([-2,2])
axes0.set_title('LINE PLOT')
plt.show()
```

```python
fig0 = plt.figure(figsize=(10,1))
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x,y,color='purple',linestyle='-')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('LINE PLOT')
plt.show()
```

```python
fig0 = plt.figure()
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x,y,color='purple',linestyle='-.',label='$y = sin(x)$')
axes0.plot(x,z,color='green',linestyle='-.', label='$y = cos(x)$')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('LINE PLOT')
axes0.legend()
plt.show()
```

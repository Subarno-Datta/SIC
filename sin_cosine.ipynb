```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)
```

```python
plt.plot(x, y_sin, color='Green', label='sin(x)')
plt.plot(x, y_cos, color='Blue', label='cos(x)')
plt.grid()
plt.xlabel('values')
plt.ylabel('sin(x) => Green, cos(x) => Blue')
plt.legend()
plt.show()
```

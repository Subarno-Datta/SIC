```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x = np.linspace(-1000, 1000, 2001)
```

```python
g = lambda x: x**2
f = lambda h: 3*h+1
def t(h):
    return (math.sqrt(h) - 1)/3
```

```python
# plt.plot(x, f(g(x)), label='f(g(x)) = $3x^2 + 1$')
plt.plot(x, g(f(x)), label='g(f(x)) = $(3x + 1)^2$')
plt.plot(x, x, linestyle='--', label='f(x) = x')
plt.legend()
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = 2*x + 3
plt.scatter(x, y)
plt.plot(x, y, label='f(x)=2x + 3')
plt.title('Linear function')
plt.legend()
plt.grid()
plt.show()
```

```python
my_dict = {
    'Size': ['L']*2 + ['M']*3 + ['S']*4,
    'Type': ['A']*3 + ['B']*2 + ['A']*2 + ['B']*2,
    'Location': ['L1']*3 + ['L2']*2 + ['L1', 'L2', 'L2', 'L1'],
    'A': [1, 2, 2, 3, 3, 4, 5, 6, 7],
    'B': [2, 4, 5, 5, 6, 6, 8, 9, 9]
}
```

```python
df = pd.DataFrame(my_dict)
```

```python
df
```

```python
dfr = pd.pivot_table(df, index=['Size', 'Type'], columns='Location', values='B')
```

```python
dfr
```

```python
pd.pivot_table(df, index=['Size', 'Type'], columns='Location', values='B', fill_value=0)
```

```python
pd.pivot_table(df, index=['Size', 'Type'], columns='Location', values='B', aggfunc='mean', fill_value=0)
```

```python
pd.pivot_table(df, index=['Size', 'Type'], values=['A','B'], aggfunc='mean')
```

```python
df.groupby(['Size', 'Type'])[['A', 'B']].median()
```

```python
pd.pivot_table(df, index=['Size', 'Type'], columns='Location', values=['A','B'], aggfunc={'A':'max', 'B': 'min'})
```

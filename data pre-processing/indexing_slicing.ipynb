```python
def is_prime(n: int):
    if (n <= 1):
        return False
    for i in range(2, 1+int(n**0.5)):
        if n%i==0:
            return False
    return True
```

```python
a3 = np.arange(1, 101, 1)
v_is_prime = np.vectorize(is_prime)
```

```python
a3[v_is_prime(a3)]
```

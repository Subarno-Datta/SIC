```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Hp', 'Db', 'Js', 'Jm', 'Jc'],
    'Gender': ['M', 'M', 'M', 'M', 'F'],
    'Age': [23, 31, 22, 36, 30]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    'Name':['Js','Adb','Jr','Jc'],
    'Position':['Intern','Team Lead','Manager','Manager'],
    'Wage':[25000,75000,90000,70000]
})
print(df2)
```

```python
help(pd.merge)
```

```python
pd.merge(df1,df2,on='Name')
```

```python
pd.merge(df1,df2,left_on='Name',right_on='Name',how='inner')
```

```python
pd.merge(df1,df2,left_on='Name',right_on='Name',how='left')
```

```python
pd.merge(df1,df2,left_on='Name',right_on='Name',how='right')
```

```python
pd.merge(df1,df2,left_on='Name',right_on='Name',how='outer')
```

```python
help(pd.concat)
```

```python
pd.concat([df1,df2],sort=True)
```

```python
pd.concat([df1,df2],axis=1,sort=True)
```

```python
df1=pd.DataFrame({'a':['a0','a1','a2','a3'],
                  'b':['b0','b1','b2','b3'],
                  'c':['c0','c1','c2','c3']},
                 index=[0,1,2,3])
df1
                  
```

```python
pd.concat([df1,df2],ignore_index=True)   #vertical
```

```python
pd.concat([df1,df2],axis=1) #horizontal
```

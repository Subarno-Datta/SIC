```python
ai = {1, 2, 3, 5}
ml = {2, 3, 4}
ds = {3, 5, 6}
```

```python
atleast_two = (ai & ml) | (ai & ds) | (ml & ds)
print(atleast_two)
```

```python
only_ai = ai - (ml | ds)
only_ds = ds - (ml | ai)
only_ml = ml - (ai | ds)
```

```python
print(only_ai)
print(only_ml)
print(only_ds)
```

```python
predicted = {"mail1", "mail2", "mail3"}
actual = {"mail2", "mail3", "mail4"}

true_positive = predicted & actual
false_positive = predicted - actual
false_negative = actual - predicted

print("true positive ", true_positive)
print("false positive ", false_positive)
print("false negative ", false_negative)
```

```python
visited_users = {"Alice", "Bob", "Charlie", "David", "Eva"}
purchased_users = {"Charlie", "Eva", "Frank", "George"}
```

```python
common_users = visited_users & purchased_users
print(common_users)
```

```python
all_users = visited_users | purchased_users
print(all_users)
```

```python
non_buyers = visited_users - purchased_users
print(non_buyers)
```

```python
new_buyers = purchased_users - visited_users
print(new_buyers)
```

```python
predicted = visited_users
actual = purchased_users
```

```python
true_positive = predicted & actual
print(true_positive)
```

```python
false_positive = predicted - actual
```

```python
print(false_positive)
```
